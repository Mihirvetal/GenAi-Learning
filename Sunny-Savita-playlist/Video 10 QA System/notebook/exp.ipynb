```python
import os
import socket
from dotenv import load_dotenv
from llama_index.core import SimpleDirectoryReader, VectorStoreIndex, Settings
from llama_index.embeddings.google_genai import GoogleGenAIEmbedding
from llama_index.llms.google_genai import GoogleGenAI

# --- DIAGNOSTIC: Prove Python can see the internet ---
try:
    socket.gethostbyname("generativelanguage.googleapis.com")
    print("🌐 Python network stack is healthy and can see Google's servers!")
except socket.gaierror:
    print("❌ NETWORK ERROR: Python still can't see the internet. Please fully restart the Anaconda Prompt.")

# 1. Load keys safely WITHOUT clearing the system environment
load_dotenv(override=True)
api_key = os.getenv("GOOGLE_API_KEY")

if not api_key:
    print("❌ ERROR: 'GOOGLE_API_KEY' not found.")
else:
    print("✅ Key found! Initializing models...")
    
    # 2. Parse the developer profile data
    document = SimpleDirectoryReader('../Data')
    doc = document.load_data()

    # 3. Configure global models
    Settings.embed_model = GoogleGenAIEmbedding(model_name="gemini-embedding-001", api_key=api_key)
    Settings.llm = GoogleGenAI(model="gemini-2.5-flash", api_key=api_key)

    # 4. Configure global chunking rules
    Settings.chunk_size = 1024
    Settings.chunk_overlap = 200

    # 5. Generate the index vectors
    print("Generating embeddings and constructing vector storage maps...")
    index = VectorStoreIndex.from_documents(doc)
    print("🚀 SUCCESS: Index built smoothly!")
```

```python
import nest_asyncio
from IPython.display import Markdown, display

# 1. Apply the patch to fix Jupyter's async loop conflict
nest_asyncio.apply()

# 2. Turn your index into a query engine
query_engine = index.as_query_engine()

# 3. Ask your question
question = "What are the core skills and strengths of this candidate?"
print(f"Asking: {question}\n")

# 4. Generate the response
response = query_engine.query(question)

# 5. Print it beautifully
display(Markdown(f"**Answer:** {response}"))
```

```python
import os
import nest_asyncio
from dotenv import load_dotenv
from llama_index.core import StorageContext, load_index_from_storage, Settings
from llama_index.embeddings.google_genai import GoogleGenAIEmbedding
from llama_index.llms.google_genai import GoogleGenAI

# 1. Standard Setup
nest_asyncio.apply()
load_dotenv(override=True)
api_key = os.getenv("GOOGLE_API_KEY")

# 2. Re-initialize your models (Settings must always be loaded)
Settings.embed_model = GoogleGenAIEmbedding(model_name="gemini-embedding-001", api_key=api_key)
Settings.llm = GoogleGenAI(model="gemini-2.5-flash", api_key=api_key)

# 3. LOAD FROM DISK (Skipping SimpleDirectoryReader completely!)
print("Loading saved vector database from disk...")
storage_context = StorageContext.from_defaults(persist_dir="./storage")
index = load_index_from_storage(storage_context)
print("✅ Index loaded instantly! Ready to query.")

```

```python
# 4. Create your query engine and ask questions!
query_engine = index.as_query_engine()
response = query_engine.query("What are the strengths i have?")
# print(response)
display(Markdown(f"**Answer:** {response}"))
```
